## **Importing the necessary Libaries.**

```python
import pandas as pd
```

### **1. Putting all file path together.**

```python
files = [
    "ACS_2018.xlsx",
    "ACS_2019.xlsx",
    "ACS_2020.xlsx",
    "ACS_2021_N.xlsx",
    "ACS_2022.xlsx"
]
```

### **2. Reading and taking random 10% sample from each year due to the large amount of data each dataset has.**

Due to the large size of ACS microdata, a 10% weighted sample (using the PERWT variable) was drawn from each year to reduce computational burden while maintaining population-level representativeness.

```python
sampled_dfs = []

for file in files:
    df = pd.read_excel(file)

    df['PERWT'] = pd.to_numeric(df['PERWT'], errors='coerce')
    df = df[df['PERWT'].notna() & (df['PERWT'] > 0)]

    df_sample = df.sample(frac=0.10, weights ="PERWT", random_state=42)

    sampled_dfs.append(df_sample)
    print(f"Sampled {file}: {len(df_sample):,} rows")
```

### **3. Putting all the years together to form a new dataset.**

```python
us_df = pd.concat(sampled_dfs, ignore_index=True)
```
