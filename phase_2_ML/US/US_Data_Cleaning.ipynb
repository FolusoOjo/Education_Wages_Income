### **Dataset Columns:**

* **YEAR**: The year in which the data was recorded (ranging from 2018–2022).

  * Corresponds to the ACS survey year.


* **WEIGHT**: Person-level survey weight (PERWT in IPUMS ACS).

  * Used to scale each individual record so the sample represents the U.S. population accurately.
  * All descriptive statistics should apply this weight to maintain representativeness.


* **TOTAL_INCOME**: Total personal income (INCTOT).

  * Includes wage income, business income, investment income, retirement income, and other income sources.
  * Reported in U.S. dollars (USD).


* **WAGE_INCOME**: Wage and salary income (INCWAGE).

  * Income received from employers before deductions.
  * Reported in U.S. dollars (USD).


* **TOTAL_EARNINGS**: Total earned income (INCEARN).

  * Includes wage/salary income and self-employment income.
  * Reported in U.S. dollars (USD).


* **STATE**: State of residence (STATEFIP).

  * Identifies the U.S. state where the respondent resides.
  * Based on Federal Information Processing Standards (FIPS) codes.


* **REGION**: Study region classification (constructed variable).

  * California
  * Rest of United States
    * Study Design Note:
        In Canada, Ontario was selected as the focal province.
        In the United States, California was chosen because it is comparable to Ontario in terms of population size, economic scale, immigrant concentration and labor market diversity.

This allows for more meaningful subnational comparison rather than full-country aggregation.



* **GENDER**: Biological sex of the respondent (SEX).

  * 1 = Male
  * 2 = Female


* **IMMIGRANT_STATUS**: Immigrant classification derived from birthplace (BPL).

  * Born in US = Born in the United States.
  * Immigrant = Born outside the United States.
  * Constructed variable to harmonize with Canadian immigrant status definition.


* **EDUCATION**: Educational attainment level.

  * Less than high school
  * Graduated high school or partial postsecondary
  * Non-university postsecondary certificate/diploma
  * University degree or certificate

Creating the imigrant status column from the Birthplace code column.

```python
df['immigrant_status'] = df['birthplace_code'].apply(
    lambda x: "Born in US" if x in state_codes else "Immigrant"
)
```

```python
df['immigrant_status'].value_counts()
```

Creating the Education column from the Education_detail column.

```python
def map_us_education(code):
    if pd.isna(code) or code in [0, 99, 999]:
        return np.nan
    elif code <= 61:
        return "Less than high school"
    elif 62 <= code <= 64:
        return "High school diploma"
    elif 65 <= code <= 90:
        return "Postsecondary certificate or diploma"
    elif code >= 100:
        return "University degree"
    else:
        return np.nan
```

```python
df["education"] = df["education_detail"].apply(map_us_education)
```

In the total income column, we are keeping the negative values as a result of information from the documentation as this shows us that there was real losses and it's very much valid.

Also, we noticed high values like 999999 and from the documentation, we've also noticed that it's not a real income so we have to drop it.



```python
df['total_income'] = df['total_income'].replace([9999999, 9999998], pd.NA)
```

```python
df['wage_income'] = df['wage_income'].replace([999999, 999998], pd.NA)
```

```python
df['total_income'] = pd.to_numeric(df['total_income'], errors='coerce')
df['wage_income'] = pd.to_numeric(df['wage_income'], errors='coerce')
df['total_earnings'] = pd.to_numeric(df['total_earnings'], errors='coerce')
```

```python
cols_to_drop = ['gender_code', 'state_code', 'age', 'marital_status_code', 'education_code', 'education_detail', 'birthplace_code', 'birthplace_detail']

df = df.drop(columns=cols_to_drop)
```

Dropping missing columns in the Education detail and Missing Income.

```python
df = df.dropna(subset=['education', 'total_income'])
```
