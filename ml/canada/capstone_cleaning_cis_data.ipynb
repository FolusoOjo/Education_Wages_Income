## Statistics Canada, Canadian Income Survey (CIS) Public Use Microdata File (PUMF), 2018–2022 editions.

### Dataset Columns:

* **YEAR**: The year in which the data was recorded (ranging from 2018-2022).


* **PUMFID**: Public Use Microdata File (PUMF) household identifier.

  * A unique numeric ID assigned to each household.

* **PERSONID**: Unique individual identifier within the dataset.

  * Each person in a household receives a distinct ID.

* **PROV**: Province or territory of residence.

  * `10 = Newfoundland and Labrador`
  * `11 = Prince Edward Island`
  * `12 = Nova Scotia`
  * `13 = New Brunswick`
  * `24 = Quebec`
  * `35 = Ontario`
  * `46 = Manitoba`
  * `47 = Saskatchewan`
  * `48 = Alberta`
  * `59 = British Columbia`
  * `60 = Yukon`
  * `61 = Northwest Territories`
  * `62 = Nunavut`

* **FWEIGHT**: Final survey weight.

  * Used to scale each record so the sample represents the Canadian population accurately.

* **AGEGP**: Coded `4–15`, typically representing grouped age ranges.

  * `04 = 15-17`
  * `05 = 18-24`
  * `06 = 25-29`
  * `07 = 30-34`
  * `08 = 35-39`
  * `09 = 40-44`
  * `10 = 45-49`
  * `11 = 50-54`
  * `12 = 55-59`
  * `13 = 60-64`
  * `14 = 65-69`
  * `15 = 70+`

* **SEX**: Sex of respondent.

  * `1 = Male`
  * `2 = Female`

* **MARSTP**: Marital status (person-level).

  * `1 = Married`
  * `2 = Common-law`
  * `3 = Separated, divorced or widowed`
  * `4 = Single (never married)`

* **HLEV2G**: Highest level of education attained.

  * `1 = Less than high school`
  * `2 = Graduated high school or partial postsecondary`
  * `3 = Non-university postsecondary certificate/diploma`
  * `4 = University degree or certificate`
  * `9 = Not stated`

* **IMMST**: Immigration status.

  * `1 = landed immigrant`
  * `2 = non-immigrant (born in Canada)`
  * `9 = Not stated / missing`

* **VISMINFL**: Visible minority flag, available only in the 2021 and 2022 datasets.

  * `1 = Visible minority`
  * `2 = Not a visible minority`
  * `9 = Not stated / missing`

* **EARNG**: Total employment earnings (including wages, salaries, and self-employment income).

  * Reported in Canadian dollars.

* **WGSAL**: Wages and salaries only.

  * A subset of total earnings.

* **TTINC**: Total personal income from all sources (before taxes).

  * Includes wages, self-employment, government transfers, investment income, etc.

## **1.2.** Data Preprocessing and Cleaning:

**1.2.1.** Read the dataset:

```python
cis_data = pd.read_csv('./cis_data_raw.csv')
cis_data.head()
```

**1.2.2.** Changing the Column names into something readable.

```python
cis_data.rename(columns={
    "YEAR": "year",
    "PUMFID": "household_id",
    "PERSONID": "person_id",
    "PROV": "geography",
    "FWEIGHT": "weight",
    "AGEGP": "age_group",
    "SEX": "gender",
    "MARSTP": "marital_status",
    "HLEV2G": "education",
    "IMMST": "immigrant_status",
    "VISMINFL": "visible_minority",
    "EARNG": "earnings",
    "WGSAL": "wages_salary",
    "TTINC": "total_income"
}, inplace=True)
```

**1.2.2.1.** Confirming that the column names are readable.

```python
cis_data.head()
```

**1.2.3.** Checking for detailed information about the data.

```python
cis_data.info()
```

**1.2.4.** Checking for missing values.

```python
cis_data_missing= cis_data.isnull().sum()
cis_data_missing
```

**1.2.5.** Creating a region column from Geography column.

**1.2.5.1.** Mapping the correct values for each province in the Geography column.

```python
province_map = {
    10: "Newfoundland and Labrador",
    11: "Prince Edward Island",
    12: "Nova Scotia",
    13: "New Brunswick",
    24: "Quebec",
    35: "Ontario",
    46: "Manitoba",
    47: "Saskatchewan",
    48: "Alberta",
    59: "British Columbia",
    60: "Yukon",
    61: "Northwest Territories",
    62: "Nunavut"
}

cis_data['geography'] = cis_data['geography'].map(province_map)
```

```python
cis_data = cis_data.dropna(subset=['geography'])
```

**1.2.5.2.** Creating a region column from geography column.

```python
cis_data['region'] = cis_data['geography'].apply(
    lambda x: 'Ontario' if x == 'Ontario' else 'Rest of Canada'
)
```

**1.2.6.** Mapping the numerical values in the 'Education' column to descriptive categories

**1.2.6.1.** Mapping the correct values for each education status.

```python
education_map = {
    1: "Less than high school",
    2: "High school diploma",
    3: "Postsecondary certificate or diploma",
    4: "University degree",
    9: "Not stated"
}

cis_data['education'] = cis_data['education'].map(education_map)
```

**1.2.6.2.** Dropping the Missing entries.

```python
cis_data = cis_data[cis_data['education'] != "Not stated"]
```

**1.2.7.** Mapping the numerical values in the 'Gender' column to its descriptive category.

```python
gender_map = {
    1: "Male",
    2: "Female"
}
cis_data['gender'] = cis_data['gender'].map(gender_map)
```

**1.2.8.** Mapping the numerical values in the 'Immigrant' column to its descriptive category.

**1.2.8.1.** Mapping the correct values for each Immigration status.

```python
immigrant_map = {
    1: "Immigrant",
    2: "Born in Canada (non-immigrant)",
    9: "Not stated / missing"
}

cis_data['immigrant_status'] = cis_data['immigrant_status'].map(immigrant_map)
```

**1.2.8.2.** Dropping the Missing entries.

```python
cis_data = cis_data[cis_data['immigrant_status'] != "Not stated / missing"]
```

**1.2.9.** Dropping the Age Group column.

```python
cis_data = cis_data.drop(columns=['age_group'])
```

**1.2.10.** Dropping the visible_minority column.

This column was removed because it does not directly address our analytical research questions. It also contains a significant number of missing (NaN) values, which could negatively affect the quality and reliability of the analysis if retained.

```python
cis_data = cis_data.drop(columns=['visible_minority'])
```

**1.2.11.** Dropping the marital_status, household_id and person_id column.

```python
cis_data = cis_data.drop(columns=['marital_status', 'household_id', 'person_id'])
```

**1.2.12.** Removing negative income values.

```python
cis_data = cis_data[cis_data['total_income'] >= 0]
```

**1.2.13.** Reseting the index of the dataset

```python
cis_data.reset_index(drop=True, inplace=True)
```
